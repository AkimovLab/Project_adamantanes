```python
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_pattern = "../for_UV_vis_spectra/step_*_spectrum.txt"  # Like, totol_steps_3000_spectrum.txt
file_list = [f for f in os.listdir() if f.startswith("step_") and f.endswith("_spectrum.txt")]

print("file found：", file_list)

x_values_list = []  
y_values_list = []  

for file_name in file_list:
    try:
        print(f"file processing: {file_name}")
        data = np.loadtxt(file_name)
        x_values_list.append(data[:, 0])  
        y_values_list.append(data[:, 1])  
    except FileNotFoundError:
        print(f"{file_name} file no found")
    except ValueError:
        print(f"{file_name} file format wrong")

# Make sure the data list is not empty
if x_values_list and y_values_list:
    # The x-values and y-values of all files are summed and averaged.
    x_values_avg = np.mean(x_values_list, axis=0)  
    y_values_avg = np.mean(y_values_list, axis=0)  

    # Normalize y_values_avg so that its maximum value is 1
    y_values_avg /= np.max(y_values_avg)  

    
    average_data = np.column_stack((x_values_avg, y_values_avg))
    np.savetxt('average_spectrum_data.txt', average_data, fmt='%.6f', header='X_avg Y_avg', comments='')

   
    plt.figure(figsize=(10, 6))
    plt.plot(x_values_avg, y_values_avg, label='Normalized Average Spectrum', color='blue')
    plt.xlabel('Energy (eV)', fontsize=14)  
    plt.ylabel('Intensity', fontsize=14)  
    plt.title('Normalized Average Spectrum Curve', fontsize=16)
    plt.legend(fontsize=12)
    plt.tight_layout()
    plt.savefig('average_spectrum_curve001.png', dpi=300)  

    
    plot_data = np.column_stack((x_values_avg, y_values_avg))
    np.savetxt('average_spectrum_data.txt', plot_data, fmt='%.6f', header='X Y', comments='')

    plt.show()
else:
    print("No useful files found.")
```

```python
file_paths = [ 'average_spectrum_data.txt']


y_offset = 0.0


plt.figure(figsize=(10, 8))


colors = [ '#A8DADC']
legend_labels = [ 'adamantane']


peak_points_64_67 = []
peak_points_70_73 = []


height_factor = 1.0


distance_between_mountains = 1000 

for i, (file_path, color, label) in enumerate(zip(file_paths, colors, legend_labels)):
    try:
 
        data = np.genfromtxt(file_path, invalid_raise=False)
        if data.ndim < 2 or data.shape[1] < 2:
            print(f"Skipping file {file_path}: insufficient data columns.")
            continue
    except ValueError as e:
        print(f"Error loading {file_path}: {e}")
        continue


    x = data[:, 0]
    y = data[:, 1] * height_factor  

    

    mask_64_67 = (x >= 6.0) & (x <= 6.7)
    x_range_64_67 = x[mask_64_67]
    y_range_64_67 = y[mask_64_67]

    if len(x_range_64_67) > 0:
        peak_x_64_67 = x_range_64_67[np.argmax(y_range_64_67)]
        peak_y_64_67 = y_range_64_67.max()
  
        peak_points_64_67.append((peak_x_64_67, peak_y_64_67 + y_offset))


    mask_70_73 = (x >= 6.8) & (x <= 7.3)
    x_range_70_73 = x[mask_70_73]
    y_range_70_73 = y[mask_70_73]

    if len(x_range_70_73) > 0:
        peak_x_70_73 = x_range_70_73[np.argmax(y_range_70_73)]
        peak_y_70_73 = y_range_70_73.max()
     
        peak_points_70_73.append((peak_x_70_73, peak_y_70_73 + y_offset))

  
    plt.fill_between(x, y_offset, y + y_offset, label=label, alpha=0.7, zorder=len(file_paths) - file_paths.index(file_path), color=color)


    y_offset += distance_between_mountains

 
    height_factor += 0.1  #


if len(peak_points_64_67) > 1:
    points_64_67 = np.array(peak_points_64_67)
    plt.plot(points_64_67[:, 0], points_64_67[:, 1], 'r--', linewidth=2, zorder=len(file_paths) + 1)


print("Red points (6.4–6.7 eV range) x-axis values:", [p[0] for p in peak_points_64_67])


for peak_x, peak_y in peak_points_64_67:
    plt.plot(peak_x, peak_y, 'ro', markersize=10, zorder=len(file_paths) + 2)


if peak_points_64_67:
    max_peak_x, max_peak_y = peak_points_64_67[-1]
    plt.text(max_peak_x - 0.05, max_peak_y + 0, '3s', color='black', fontsize=22, ha='center', zorder=len(file_paths) + 3)


if len(peak_points_70_73) > 1:
    points_70_73 = np.array(peak_points_70_73)
    plt.plot(points_70_73[:, 0], points_70_73[:, 1], color='orange', linestyle='--', linewidth=2, zorder=len(file_paths) + 1)


print("Orange points (7.0–7.3 eV range) x-axis values:", [p[0] for p in peak_points_70_73])


for peak_x, peak_y in peak_points_70_73:
    plt.plot(peak_x, peak_y, 'o', color='orange', markersize=10, zorder=len(file_paths) + 2)


if peak_points_70_73:
    max_peak_x_70_73, max_peak_y_70_73 = peak_points_70_73[-1]
    plt.text(max_peak_x_70_73 - 0.05, max_peak_y_70_73 + 0, '3p', color='black', fontsize=22, ha='center', zorder=len(file_paths) + 3)

plt.xlim(6.2, 8.2)
plt.xticks(np.arange(6.2, 8.1, 0.3), fontsize=24)
plt.title('UV-Vis Spectra Analysis Plot', fontsize=24)
plt.xlabel('Excitation energy (eV)', fontsize=24)
plt.ylabel('$\\varepsilon$(L mol$^{-1}$ cm$^{-1}$)', fontsize=24)
plt.xticks(fontsize=24)


plt.legend(fontsize=24)


plt.yticks([])


plt.tight_layout()  
plt.savefig('AD_300K.png', dpi=600)  
plt.show()
```
