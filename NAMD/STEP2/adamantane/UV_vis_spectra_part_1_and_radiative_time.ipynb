```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
import re
```

### UV–Vis Spectrum Extraction

This code extracts excitation energies and oscillator strengths from TDDFPT log files matching:

`all_logfiles/step_*.log`

The discrete transitions are converted into continuous spectra using Gaussian broadening:

$$
I(E)=\sum_i f_i
\exp\left[
-\frac{(E-E_i)^2}{2\sigma^2}
\right]
$$

with:

$$
\sigma=\frac{\mathrm{FWHM}}{2.35482}
$$

Here, the FWHM is set to $0.1\ \mathrm{eV}$.

The resulting spectral data are saved in:

`for_UV_vis_spectra`

Each output file contains two columns:

`Energy (eV)    Intensity (a.u.)`


```python
# Function to parse spectrum data from a log file
def parse_spectrum_data_from_log(file_path):
    energy_levels, intensities = [], []
    with open(file_path, 'r') as file:
        for line in file:
            if line.startswith(' TDDFPT|') and len(line.split()) >= 7:
                parts = line.split()
                try:
                    energy_levels.append(float(parts[2]))  # Excitation energy (eV)
                    intensities.append(float(parts[-1]))  # Oscillator strength
                except ValueError:
                    continue
    return np.array(energy_levels), np.array(intensities)

# Function to calculate spectrum with Gaussian broadening
def gaussian_broadening(energy_levels, intensities, fwhm, num_points=1000):
    if len(energy_levels) == 0 or len(intensities) == 0:
        raise ValueError("Energy levels or intensities array is empty. Check input data.")
    x = np.linspace(np.min(energy_levels) - 5 * fwhm, np.max(energy_levels) + 5 * fwhm, num_points)
    y = np.sum([intensity * np.exp(-((x - energy) ** 2) / (2 * (fwhm / 2.35482) ** 2)) for energy, intensity in zip(energy_levels, intensities)], axis=0)
    return x, y

# Function to generate and save spectrum for each log file
def generate_and_save_spectra(log_file_pattern, output_folder, fwhm=0.3, num_points=1000):
    # Create output folder if it doesn't exist
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
    
    log_files = glob.glob(log_file_pattern)
    if not log_files:
        raise FileNotFoundError(f"No files matched the pattern: {log_file_pattern}")

    for log_file in log_files:
        print(f"Processing: {log_file}")
        energy_levels, intensities = parse_spectrum_data_from_log(log_file)

        # Generate Gaussian broadened spectrum
        x, y = gaussian_broadening(energy_levels, intensities, fwhm, num_points)

        # Create output file path
        base_name = os.path.basename(log_file).replace('.log', '_spectrum.txt')
        output_file_path = os.path.join(output_folder, base_name)

        # Save spectrum data
        np.savetxt(output_file_path, np.column_stack((x, y)), header="Energy (eV)\tIntensity (a.u.)")
        print(f"Spectrum data saved to: {output_file_path}")

# Example usage:
log_file_pattern = "all_logfiles/step_*.log"
output_folder = "for_UV_vis_spectra"
generate_and_save_spectra(log_file_pattern, output_folder, fwhm=0.1, num_points=1000)
```

### Radiative Lifetime Calculation

This code extracts the excitation energy and oscillator strength of the selected TDDFPT state from each file matching:

`all_logfiles/step_*.log`

For each trajectory step, the radiative lifetime is calculated as:

$$
\tau_{\mathrm{rad}} =
\frac{46.7}{2E^2f}
$$

where $E$ is the excitation energy in eV and $f$ is the oscillator strength.

The script records the following quantities for each step:

`Step    Energy (eV)    Oscillator strength    Radiative lifetime(ns)`

Invalid entries, including missing data and zero oscillator strength, are excluded.

The calculated radiative lifetimes are summarized using the number of valid samples, mean, standard deviation, standard error, 95% confidence interval, minimum, and maximum values.

The 95% confidence interval is estimated as:

$$
\mathrm{CI}_{95\%} =
1.96\frac{s}{\sqrt{N}}
$$

where $s$ is the sample standard deviation and $N$ is the number of valid trajectory steps.



```python
def extract_state_properties(filename, state=1, take="last"):
    last_energy = None
    last_osc = None

    with open(filename, "r") as f:
        for line in f:
            s = line.lstrip()
            if not s.startswith("TDDFPT|"):
                continue

            parts = s.split()
            if len(parts) < 4:
                continue

            try:
                sid = int(parts[1])
            except ValueError:
                continue

            if sid == state:
                try:
                    energy = float(parts[2])      # eV
                    osc = float(parts[-1])        # f
                except ValueError:
                    continue

                if take == "first":
                    return energy, osc

                last_energy = energy
                last_osc = osc

    if last_energy is not None:
        return last_energy, last_osc
    else:
        return None, None


def extract_step_number(path):
    m = re.search(r"step_(\d+)\.log", str(path).replace("\\", "/"))
    return int(m.group(1)) if m else None


# ============================
# 主程序
# ============================
records = []
taus = []

for fname in glob.glob("all_logfiles/step_*.log"):
    step = extract_step_number(fname)
    if step is None:
        continue

    energy, osc = extract_state_properties(fname, state=1, take="last")

    if energy is None or osc is None or osc == 0:
        print(f"Warning: invalid data in {fname}")
        continue

    # ===== 计算 τ_rad =====
    tau = 46.7 / (2 * (energy**2) * osc)

    records.append((step, energy, osc, tau))
    taus.append(tau)

records.sort(key=lambda x: x[0])
taus = np.array(taus, dtype=float)

if len(taus) == 0:
    raise RuntimeError("No valid τ_rad computed.")


print("\n# step    Energy(eV)    f        tau_rad")
for step, energy, osc, tau in records:
    print(f"{step:6d}   {energy:10.5f}   {osc:.6e}   {tau:.6e}")


print("\n# step    Energy(eV)    f        tau_rad")
for step, energy, osc, tau in records:
    print(f"{step:6d}   {energy:10.5f}   {osc:.6e}   {tau:.6e}")

print("\n=== Radiative Lifetime Statistics ===")

N = len(taus)
mean_tau = taus.mean()

if N > 1:
    std_tau = taus.std(ddof=1)                 # SD
    se_tau = std_tau / np.sqrt(N)              # Standard Error
    ci95 = 1.96 * se_tau                       # 95% Confidence Interval
else:
    std_tau = np.nan
    se_tau = np.nan
    ci95 = np.nan

print(f"Count : {N}")
print(f"Mean  : {mean_tau:.6e}")
print(f"SD    : {std_tau:.6e}" if N > 1 else "SD    : N/A")
print(f"SE    : {se_tau:.6e}" if N > 1 else "SE    : N/A")
print(f"95% CI: ±{ci95:.6e}" if N > 1 else "95% CI: N/A")
print(f"Min   : {taus.min():.6e}")
print(f"Max   : {taus.max():.6e}")
```
